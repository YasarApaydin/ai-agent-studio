```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen3:1.7b",
    temperature=0
)

response = llm.invoke("Merhaba, bana Agent kavramını anlat.")

print(response.content)
```

```python
from langchain.tools import tool
@tool
def square_root(x: float) -> float:
    """Calculates the square root of a number."""
    return x ** 0.5
```

```python
@tool("square_root")
def tool1(x: float) -> float:
    """Calculates the square root of a number."""
    return x ** 0.5
```

```python
@tool("square_root",description="Calculates the square root of a number.")
def tool1(x: float) -> float:
    return x ** 0.5
```

```python
tool1.invoke({"x":467})
```

```python
from langchain.agents import create_agent
agent = create_agent(llm, tools=[tool1],system_prompt="You are a helpful assistant that can calculate the square root of a number.")
```

```python
from langchain.messages import HumanMessage
question = HumanMessage(content="What is the square root of 467?")
response = agent.invoke({"messages": [question]})
```
